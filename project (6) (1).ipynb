```python

import random
class Player:
    def __init__(self, name):
        self.name = name
        self.health = 100
        self.oxygen = 100
        self.food = 100

    def show_status(self):
        print("\n--- PLAYER STATUS ---")
        print("Name:", self.name)
        print("Health:", self.health)
        print("Oxygen:", self.oxygen)
        print("Food:", self.food)

    def consume_resources(self):
        self.oxygen -= 10
        self.food -= 10

    def is_alive(self):
        return (
            self.health > 0
            and self.oxygen > 0
            and self.food > 0
        )


class Game:
    def __init__(self, player):
        self.player = player
        self.turn = 0
        self.max_turns = 10

    def random_event(self):
        event = random.choice([
            "oxygen",
            "food",
            "storm",
            "nothing"
        ])

        if event == "oxygen":
            self.player.oxygen += 15
            print("You found an oxygen supply! +15 oxygen")

        elif event == "food":
            self.player.food += 15
            print("You found food! +15 food")

        elif event == "storm":
            self.player.health -= 20
            print("A Mars storm damaged your suit! -20 health")

        else:
            print("Nothing happened today.")

    def play_turn(self):
        self.turn += 1

        print(f"\n========== TURN {self.turn} ==========")

        self.player.consume_resources()

        self.random_event()

        self.player.show_status()

    def start(self):
        print("\nWelcome to THE LONELY ASTRONAUT!")
        print("Your mission: Survive 10 turns on Mars.")

        while (
            self.turn < self.max_turns
            and self.player.is_alive()
        ):
            input("\nPress Enter to continue...")
            self.play_turn()

        if self.player.is_alive():
            print("\nCONGRATULATIONS!")
            print("You survived 10 turns on Mars!")
        else:
            print("\nGAME OVER!")
            print("You did not survive on Mars.")


# Create player object
name = input("Enter astronaut name: ")
player1 = Player(name)
# player1 = Player("Abubakar")
# player2 = Player("Ali")
# player3 = Player("Ahmed")

# Create game object
game = Game(player1)

# Start game
game.start()
```
